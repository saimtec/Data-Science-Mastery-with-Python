```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df= sns.load_dataset('penguins')
df
```

```python
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df)
plt.show()
```

```python
# add hue parameter for colour coding
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex')
plt.show()
```

```python
# add style parameter for different markers and size parameter for point size
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='species' , s=100)
plt.show()
```

```python
# add palette parameter for custom colour palette
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='species' , s=100 , palette='rocket_r') 
plt.show()
```

```python
# specify markers for styles
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='sex' , s=100 , palette='rocket_r' , markers=['o','>']) 
plt.show()
```

```python
# adding grid and style
sns.set_style("whitegrid")
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='sex' , s=100 , palette='rocket_r' , markers=['o','>']) 
plt.show()
```

```python
sns.set_style("dark")
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='sex' , s=100 , palette='rocket_r' , markers=['o','>']) 
plt.show()
```

```python
# final plot with despine for removing top and right spines (spine=border line of the plot)
sns.set_style("whitegrid")
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='island' , style='sex' , s=100 , palette='rocket_r' , markers=['o','>' ,'^']) 
sns.despine()
plt.show()
```

```python
# final plot with context for adjusting font sizes
sns.set_style("ticks")
sns.set_context("notebook", font_scale=1.2)
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='sex' , s=100 , palette='rocket_r' , markers=['o','>']) 
plt.show()
```

```python
# final plot for paper can also use talk or poster
sns.set_style("ticks")
sns.set_context("paper")
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='sex' , s=100 , palette='rocket_r' , markers=['o','>']) 
plt.show()
```

<br><br><br>

# Strip Plot
* strip plot useful for categorical scatter plots

```python
# Example of strip plot
#dodge parameter to separate points based on hue variable
#set2 is a qualitative colour palette
sns.stripplot(x='species' , y= 'body_mass_g', data=df , hue='island' , dodge=True , palette='Set2')
plt.show()
```

<br><br>

# Swarm Ploot
* swarm plot use for better separation of points

```python
sns.swarmplot(x='species' , y= 'body_mass_g', data=df , hue='island' , dodge=True)
plt.show()
```

<br><br>

# Histogram

```python
#kde=True adds a kernel density estimate curve and is useful for visualising distribution
sns.histplot(df['flipper_length_mm'], kde=True)
plt.show()
```

```python
#cmap parameter adds a colormap for 2D histogram
sns.histplot(data=df, x='flipper_length_mm', y='body_mass_g', bins=30 ,cmap="mako")
plt.show()
```

<br><br>

# Regression plot
* useful for categorical distribution

```python
sns.regplot(x='flipper_length_mm' , y= 'body_mass_g', data=df)
plt.show()  
```

```python
# final plot for paper can also use talk or poster with title added and darkgrid style
sns.set(style="darkgrid")
sns.set_context("paper")
sns.scatterplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , hue='sex' , style='sex' , s=100 , palette='rocket_r' , markers=['o','>']) 
plt.title("Penguin Flipper Length vs Body Mass")    
plt.show()
```

<br><br>

# Jintplote
* join two plotes

```python
# joint plot to show scatter plot along with histograms of individual variables
sns.jointplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , kind='scatter')
plt.show()
```

```python
# can set own kind of plot like kde
sns.jointplot(x='flipper_length_mm' , y= 'body_mass_g', data=df , kind='kde' ,hue='sex')
plt.show()
```

# Boxplot
* show frequency distribution
* have 25% , 50% , 75% , 100% quartiles

```python
sns.boxplot(x='species' , y= 'body_mass_g', data=df ,hue='island' , palette='Set1')
plt.show()
```

<br><br>

# Violinplot
* violin plot to show distribution shape and density good for comparing multiple categories
* same as boxplot (have min , max values)
* use kde (kernal density estimation)

```python
sns.violinplot(x='species' , y= 'body_mass_g', data=df ,hue='island' , palette='Set1')
plt.show()
```

```python
# final violin plot with split parameter to show both island distributions side by side
sns.violinplot(x='species' , y= 'body_mass_g', data=df ,hue='island' , palette='Set1' , split=True) 
plt.show()
```

```python
# final violin plot with swarmplot overlayed using dodge to separate points based on hue variable
sns.violinplot(x='species' , y= 'body_mass_g', data=df ,hue='island' , palette='Set1')
sns.swarmplot(x='species' , y= 'body_mass_g', data=df , hue='island' , palette='dark' , dodge=True , color='k' , alpha=0.6)
plt.show()
```

<br><br>

# Kde (lernal density estimation)
* probability density func in density data
* shows the distribution of data
* helps to see where data is concentrated through peeks
* more balanced version of histogram

```python
# kde plot to show distribution of body mass for different species
sns.kdeplot(data=df, x='body_mass_g', hue='species', fill=False)
plt.show()
```

```python
sns.kdeplot(data=df, x='body_mass_g',  hue='species', fill=True)
plt.show()
```

```python
# final kde plot showing 2D distribution of flipper length and body mass with colormap
sns.kdeplot(data=df, x='flipper_length_mm', y='body_mass_g', cmap="mako", fill=True)
plt.show()
```
