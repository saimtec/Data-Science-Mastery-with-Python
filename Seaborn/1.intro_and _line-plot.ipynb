# Seaborn
* Seaborn is a library for making statistical graphs in python
* interrelated to matplotlib and pandas
* different categories of plots in seaborn 
 
 **Relational = plots used to understand relation between two variables**

 **Categprical = plot deal with categorical plots**
 
 **Distributional = used to examone univariate and bivariate dostribution**

 Seaborn divides plots into 
 * regression plot
 * Matrix plot
 * Multi plot grid


<br><br>

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
var1 = [1,2,3,4,5,6]
var2 = [3,4,5,6,7,8]
```

```python
#matplotlib
plt.plot(var1 , var2)
plt.show()
```

```python
# seaborn
sns.lineplot(x=var1 , y = var2)
plt.show()
```

```python
# now using pandas

import pandas as pd

df =pd.DataFrame({"var1":var1,
                   "var2":var2
                })

df
sns.lineplot(x='var1' , y= 'var2', data=df)
plt.show()
```

### unlike matplotlib seaborn adds additional details automatically 

```python
df1 = sns.load_dataset('penguins')
df1.head()
```

```python
# categprize by sex using hue
sns.lineplot(x='flipper_length_mm' , y= 'body_mass_g', data=df1 , hue='sex')
plt.show()
```

```python
# change colour palette
sns.lineplot(x='flipper_length_mm' , y= 'body_mass_g', data=df1 , hue='sex', palette='rocket')
plt.show()
```

```python
# change the markers
sns.lineplot(x='flipper_length_mm' , y= 'body_mass_g', data=df1 ,style="sex" ,hue='sex', palette='rocket', markers=['o','>'])
plt.show()
```

```python
df2 = sns.load_dataset('penguins').head(20)
df2
```

```python
sns.lineplot(x='flipper_length_mm' , y= 'body_mass_g', data=df2 ,style="sex" ,hue='sex', palette='rocket', markers=['o','>'])
plt.show()
```

```python
# adding grid and title
sns.lineplot(x='flipper_length_mm' , y= 'body_mass_g', data=df2 ,style="sex" ,hue='sex', palette='rocket', markers=['o','>'])
plt.title("Penguin Flipper Length vs Body Mass")
plt.grid(True)
plt.show()
```
